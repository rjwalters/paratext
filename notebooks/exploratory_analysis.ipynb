````python
from pathlib import Path
import pandas as pd
from paratext.dataset import read_run_records
from paratext.analysis.metrics import (
    aggregate_explicit, aggregate_implicit, comparison_table
)

explicit = list(read_run_records('../data/runs/explicit.jsonl'))
implicit = list(read_run_records('../data/runs/implicit.jsonl'))
len(explicit), len(implicit)
````

````python
explicit_agg = aggregate_explicit(explicit)
explicit_agg['per_class']
````

````python
comparison_table(explicit_agg['paired_deltas'], ('fatigue', 'rushed_or_mobile', 'frustration', 'low_bandwidth'))
````

````python
implicit_agg = aggregate_implicit(implicit)
implicit_agg['per_class']
````

### Specificity check

If the model is reading paralinguistic cues (not just reacting to noise), the `fatigue_coded` row should show a much larger fatigue delta than `random_typo_control`.

````python
deltas = explicit_agg['paired_deltas']
fatigue = deltas[deltas['field'] == 'fatigue']
fatigue.set_index('variant_class')[['delta_mean', 'n']]
````
